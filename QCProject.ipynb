```python
# ========================= X GATE =========================
def state_to_bloch_vector(state):
    a, b = state[0], state[1]
    return [2*np.real(a*np.conj(b)), 2*np.imag(b*np.conj(a)), np.abs(a)**2 - np.abs(b)**2]

qc = QuantumCircuit(1, 1)
qc.x(0)
qc.measure(0, 0)
print("Circuit Diagram (X Gate):")
display(qc.draw('mpl'))
qc_no_m = QuantumCircuit(1)
qc_no_m.x(0)
sv = Statevector.from_instruction(qc_no_m)
print("\nStatevector (X):", sv)
print("Probability Vector (X):", sv.probabilities())
bv = state_to_bloch_vector(sv.data)
print("Bloch Vector (X):", [float(x) for x in bv])
fig = plt.figure(figsize=(3,3)); b = Bloch(fig); b.add_vectors(bv); b.show(); plt.show()
sim = AerSimulator()
res = sim.run(qc, shots=1024).result()
counts = res.get_counts()
print("Measurement Counts:", counts)
```

```python
# ========================= CNOT GATE =========================
def output_qubit_bloch(state):
    # Density matrix
    rho = np.outer(state, np.conj(state))

    # Trace out qubit 0 → keep qubit 1 (output qubit)
    rho_out = np.trace(rho.reshape(2, 2, 2, 2), axis1=0, axis2=2)

    # Bloch coordinates
    x = 2 * np.real(rho_out[0, 1])
    y = 2 * np.imag(rho_out[1, 0])
    z = np.real(rho_out[0, 0] - rho_out[1, 1])

    return [x, y, z]

qc = QuantumCircuit(2, 2)
qc.x(0)
qc.cx(0, 1)

print("Circuit Diagram (CNOT Gate):")
display(qc.draw('mpl'))

qc_no_m = QuantumCircuit(2)
qc_no_m.x(0)
qc_no_m.cx(0, 1)

sv = Statevector.from_instruction(qc_no_m)

print("\nStatevector (CNOT):", sv)
print("Probability Vector (CNOT):", sv.probabilities())

# Get Bloch vector for OUTPUT qubit (qubit 1)
bv_out = output_qubit_bloch(sv.data)

print("Bloch Vector (Output Qubit):", [float(x) for x in bv_out])

# Plot Bloch sphere
fig = plt.figure(figsize=(3,3))
b = Bloch(fig)
b.add_vectors(bv_out)
b.title = "Output Qubit (Qubit 1)"
b.show()
```

```python
# ========================= CCNOT GATE (Output Qubit Bloch Only) =========================
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization.bloch import Bloch
from IPython.display import display
import matplotlib.pyplot as plt

def output_qubit_bloch_ccnot(state):
    # Density matrix
    rho = np.outer(state, np.conj(state))

    # Trace out qubits 0 and 1 → keep qubit 2 (output qubit)
    rho_out = np.trace(
        np.trace(rho.reshape(2,2,2,2,2,2), axis1=0, axis2=3),
        axis1=0, axis2=2
    )

    # Bloch coordinates
    x = 2 * np.real(rho_out[0, 1])
    y = 2 * np.imag(rho_out[1, 0])
    z = np.real(rho_out[0, 0] - rho_out[1, 1])

    return [x, y, z]

qc = QuantumCircuit(3, 3)
qc.x(0)
qc.x(1)
qc.ccx(0, 1, 2)

print("Circuit Diagram (CCNOT Gate):")
display(qc.draw('mpl'))

qc_no_m = QuantumCircuit(3)
qc_no_m.x(0)
qc_no_m.x(1)
qc_no_m.ccx(0, 1, 2)

sv = Statevector.from_instruction(qc_no_m)

print("\nStatevector (CCNOT):", sv)
print("Probability Vector (CCNOT):", sv.probabilities())

# Get Bloch vector for OUTPUT qubit (qubit 2)
bv_out = output_qubit_bloch_ccnot(sv.data)

print("Bloch Vector (Output Qubit):", [float(x) for x in bv_out])

# Plot Bloch sphere
fig = plt.figure(figsize=(3,3))
b = Bloch(fig)
b.add_vectors(bv_out)
b.title = "Output Qubit (Qubit 2)"
b.show()
```

```python
# ========================= Half Adder =========================
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

A = int(input("Enter first bit A (0 or 1): "))
B = int(input("Enter second bit B (0 or 1): "))

qc = QuantumCircuit(4, 2)

if A == 1:
    qc.x(0)
if B == 1:
    qc.x(1)

qc.cx(0, 2)
qc.cx(1, 2)
qc.ccx(0, 1, 3)

qc.measure(2, 0)
qc.measure(3, 1)

print("\nCircuit Diagram (Half Adder):")
display(qc.draw('mpl'))

sim = AerSimulator()
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=1).result()
counts = result.get_counts()

output = list(counts.keys())[0]
sum_bit = output[-1]
carry_bit = output[-2]

print(f"\nInput: A={A}, B={B}")
print(f"SUM = {sum_bit}")
print(f"CARRY = {carry_bit}")

print("\nMeasurement Results (SUM, CARRY):",counts)
```

```python
# ========================= Full Adder =========================
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

A = int(input("Enter first bit A (0 or 1): "))
B = int(input("Enter second bit B (0 or 1): "))
Cin = int(input("Enter carry-in bit Cin (0 or 1): "))

# 5 qubits:
# q0 = A
# q1 = B
# q2 = Cin
# q3 = Sum
# q4 = Carry-out
qc = QuantumCircuit(5, 2)

# Initialize inputs
if A == 1:
    qc.x(0)
if B == 1:
    qc.x(1)
if Cin == 1:
    qc.x(2)

# ---- FULL ADDER LOGIC ----

# SUM = A ⊕ B ⊕ Cin
qc.cx(0, 3)
qc.cx(1, 3)
qc.cx(2, 3)

# CARRY = (A AND B) OR (Cin AND (A ⊕ B))

# A AND B part
qc.ccx(0, 1, 4)

# Cin AND (A XOR B) part
qc.cx(0, 1)          # compute A XOR B into qubit 1 temporarily
qc.ccx(1, 2, 4)      # (A XOR B) AND Cin
qc.cx(0, 1)          # uncompute to restore B

# Measure SUM and CARRY
qc.measure(3, 0)
qc.measure(4, 1)

print("\nCircuit Diagram (Full Adder):")
display(qc.draw('mpl'))

sim = AerSimulator()
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=1).result()
counts = result.get_counts()

output = list(counts.keys())[0]
sum_bit = output[-1]
carry_bit = output[-2]

print(f"\nInput: A={A}, B={B}, Cin={Cin}")
print(f"SUM = {sum_bit}")
print(f"CARRY = {carry_bit}")

print("\nMeasurement Results (SUM, CARRY):", counts)
```

```python
# ========================= Half Subtractor =========================
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

# Inputs
A = int(input("Enter first bit A (0 or 1): "))
B = int(input("Enter second bit B (0 or 1): "))

# Half Subtractor needs 4 qubits:
# Qubits 0,1 = inputs A, B
# Qubits 2,3 = ancilla for Difference and Borrow
qc = QuantumCircuit(4, 2)

# Initialize input qubits
if A == 1:
    qc.x(0)
if B == 1:
    qc.x(1)

# Difference = A XOR B
qc.cx(0, 2)
qc.cx(1, 2)

# Borrow_out = (~A & B)
qc.x(0)            # NOT A
qc.ccx(0, 1, 3)    # (~A & B)
qc.x(0)            # Restore A

# Measure Difference and Borrow_out
qc.measure(2, 0)  # Difference
qc.measure(3, 1)  # Borrow_out

# Display circuit
print("\nCircuit Diagram (Half Subtractor):")
display(qc.draw('mpl'))

# Simulation
sim = AerSimulator()
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=1).result()
counts = result.get_counts()

# Extract results
output = list(counts.keys())[0]
diff_bit = output[-1]
borrow_bit = output[-2]

print(f"\nInput: A={A}, B={B}")
print(f"DIFFERENCE = {diff_bit}")
print(f"BORROW_OUT = {borrow_bit}")

print("\nMeasurement Results (DIFFERENCE, BORROW_OUT):", counts)
```

```python
# ========================= Full Subtractor =========================
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display

# Inputs
A = int(input("Enter first bit A (0 or 1): "))
B = int(input("Enter second bit B (0 or 1): "))
Bin = int(input("Enter Borrow in (Bin) (0 or 1): "))

# Full Subtractor needs 5 qubits:
# Qubits 0,1,2 = inputs A, B, Bin
# Qubits 3,4 = ancilla for computation of D and Bout
qc = QuantumCircuit(5, 2)

# Initialize input qubits
if A == 1:
    qc.x(0)
if B == 1:
    qc.x(1)
if Bin == 1:
    qc.x(2)

# Difference = A XOR B XOR Bin
qc.cx(0, 3)
qc.cx(1, 3)
qc.cx(2, 3)

# Borrow_out = (~A & B) | (~A & Bin) | (B & Bin)
# Step 1: compute (~A & B) => need to use X gates for NOT
qc.x(0)
qc.ccx(0, 1, 4)  # (~A & B)
qc.x(0)

# Step 2: compute (~A & Bin)
qc.x(0)
qc.ccx(0, 2, 4)  # (~A & Bin)
qc.x(0)

# Step 3: compute (B & Bin)
qc.ccx(1, 2, 4)

# Measure Difference and Borrow_out
qc.measure(3, 0)  # Difference
qc.measure(4, 1)  # Borrow_out

# Display circuit
print("\nCircuit Diagram (Full Subtractor):")
display(qc.draw('mpl'))

# Simulation
sim = AerSimulator()
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=1).result()
counts = result.get_counts()

# Extract results
output = list(counts.keys())[0]
diff_bit = output[-1]
borrow_bit = output[-2]

print(f"\nInput: A={A}, B={B}, Bin={Bin}")
print(f"DIFFERENCE = {diff_bit}")
print(f"BORROW_OUT = {borrow_bit}")

print("\nMeasurement Results (DIFFERENCE, BORROW_OUT):", counts)
```

```python
# ========================= AB⋅(CD'+EF) =========================
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from IPython.display import display

# Inputs
A = int(input("Enter A (0 or 1): "))
B = int(input("Enter B (0 or 1): "))
C = int(input("Enter C (0 or 1): "))
D = int(input("Enter D (0 or 1): "))
E = int(input("Enter E (0 or 1): "))
F = int(input("Enter F (0 or 1): "))

# Qubit allocation:
# 0:A, 1:B, 2:C, 3:D, 4:E, 5:F
# 6: AB
# 7: C D'
# 8: EF
# 9: (C D' + EF)
# 10: final output X

qc = QuantumCircuit(11, 1)

# Initialize inputs
inputs = [A, B, C, D, E, F]
for i, val in enumerate(inputs):
    if val == 1:
        qc.x(i)

# ---- Step 1: AB ----
qc.ccx(0, 1, 6)

# ---- Step 2: C D' ----
qc.x(3)              # D -> D'
qc.ccx(2, 3, 7)
qc.x(3)              # restore D

# ---- Step 3: EF ----
qc.ccx(4, 5, 8)

# ---- Step 4: OR = (C D' + EF) ----
# OR using: a + b = a XOR b XOR (a AND b)
qc.cx(7, 9)
qc.cx(8, 9)
qc.ccx(7, 8, 9)

# ---- Step 5: Final AND with AB ----
qc.ccx(6, 9, 10)

# Measure output X
qc.measure(10, 0)

# Draw circuit
print("\nCircuit Diagram:")
display(qc.draw('mpl'))

# Simulate
sim = AerSimulator()
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=1).result()
counts = result.get_counts()

# Output
output = list(counts.keys())[0]


print("\nInputs:")
print(f"A={A}, B={B}, C={C}, D={D}, E={E}, F={F}")
print(f"\nOUTPUT X = {output}")
print("\nMeasurement Counts:", counts)
```

```python
# ========================= X=AB+CDE =========================
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from IPython.display import display

# Inputs
A = int(input("Enter A (0 or 1): "))
B = int(input("Enter B (0 or 1): "))
C = int(input("Enter C (0 or 1): "))
D = int(input("Enter D (0 or 1): "))
E = int(input("Enter E (0 or 1): "))

# Qubit allocation:
# 0:A, 1:B, 2:C, 3:D, 4:E
# 5: AB
# 6: CD
# 7: CDE
# 8: Output X

qc = QuantumCircuit(9, 1)

# Initialize inputs
inputs = [A, B, C, D, E]
for i, val in enumerate(inputs):
    if val == 1:
        qc.x(i)

# ---- Step 1: AB ----
qc.ccx(0, 1, 5)

# ---- Step 2: CD ----
qc.ccx(2, 3, 6)

# ---- Step 3: CDE ----
qc.ccx(6, 4, 7)

# ---- Step 4: OR = AB + CDE ----
qc.cx(5, 8)
qc.cx(7, 8)
qc.ccx(5, 7, 8)

# Measure output
qc.measure(8, 0)

# Draw circuit
print("\nCircuit Diagram:")
display(qc.draw('mpl'))

# Simulate
sim = AerSimulator()
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=1).result()
counts = result.get_counts()

# Output
output = list(counts.keys())[0]

print("\nInputs:")
print(f"A={A}, B={B}, C={C}, D={D}, E={E}")
print(f"\nOUTPUT X = {output}")
print("\nMeasurement Counts:", counts)
```
